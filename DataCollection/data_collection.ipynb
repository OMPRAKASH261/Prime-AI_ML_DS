```python
# BeatuifulSoup

from bs4 import BeautifulSoup

with open("scraped_data/data1.html", "r") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, "lxml")
```

```python
# Methods
## -> find()  - first matching element
soup.find("h1")
soup.h1
soup.find("h3")

## -> find_all()
soup.find_all("h3")

all_h3 = soup.find_all("h3")
type(all_h3)


all_countries = []


for h3 in all_h3:
    # print(h3.get_text(strip=True))

## -> find_parent()
    name = h3.get_text(strip=True)
    # print(h3.find_parent("div")["class"])

## -> find_next()
    # print(h3.find_next("div"))
    
### css selectors:- 1. select()  2. select_one()
    # -> select()
    population = h3.find_next("div").select("span.country-population")[0].get_text(strip=True)
    # print(population)

    # -> select_one()
    population = h3.find_next("div").select_one("span.country-population").get_text(strip=True)
    # print(population)

### attributes:- 1. get()  2. .attrs
    # print(h3.attrs)

    all_countries.append([name, population])

    
### Nevigating DOM (Document Object Model):- parent,chidren,descendants,next_sibling(),previous_sibling()
    
    # print(h3.parent)
    # print(h3.children)
```

```python
all_countries
```

```python
import pandas as pd

df = pd.DataFrame(all_countries, columns=["Name", "Population"])
```

```python
## store our clean data in folder
df.to_csv("cleaned_data/data.csv", index=False)
```
