### Featrue Engineering - Data Transformation

```python
df2 = df.copy()

df2["tax"] = df2["income"].apply(lambda x : "20%" if x >= 60000 else "10%")

gender_map = {"Male" : "M", "Female" : "F", "Unknown" : "U"}
df2["gender"] = df2["gender"].map(gender_map)

df2 = df2.assign(new_income = df2["income"] * 1.1)

df2["country"] = df2["country"].replace("USA", "Unitied States")

df2.columns = ["Id", "Name", "Age", "Country", "Gender", "Income", "Tax", "New_Income"]
df2.rename(columns={"Income":"Salary"})
df2.rename(index={1:"First"})

df2.sort_values("Income")
df2 = df2.fillna(50)
# df2.sort_values("Income", ascending=False)
sorted_df = df2.sort_values(["Income","Age"])
sorted_df

sorted_df.sort_index()

#reset
sorted_df.reset_index()
sorted_df.reset_index(drop=True)

#rank
sorted_df["Ranking"] = sorted_df["Income"].rank()
sorted_df["Ranking"] = sorted_df["Income"].rank(ascending=False, method="max")
sorted_df

#reorder
sorted_df[["Id", "Name", "Age", "Gender", "Country", "Income", "Tax","Ranking", "New_Income"]]
```

#### Melt & Pivot

```python
df = pd.DataFrame({
    "country": ["USA", "USA", "India", "India"],
    "year": [2020, 2021, 2020, 2021],
    "sales": [100, 120, 90, 110],
    "profit": [20, 25, 18, 22]
})

melted_df = df.melt(
    id_vars=["country", "year"],
    value_vars=["sales", "profit"],
    var_name= "metrics",
    value_name= "value"
)

original = melted_df.pivot(
    index=["country", "year"],
    columns="metrics",
    values = "value"
)
```

#### Merging & Joining Data

```python
df_customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4],
    "name": ["Adam", "Bob", "Charlie", "Dave"]
})

df_orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104],
    "customer_id":[2, 1, 4, 5],
    "amount": [250, 120, 300, 180]
})

# pd.merge(df_customers, df_orders, on="customer_id")  #inner join
```

```python
df_customers
```

```python
df_orders
```

```python
#inner join
pd.merge(df_customers, df_orders, on="customer_id")
```

```python
#left join
pd.merge(df_customers, df_orders, on="customer_id", how="left")
```

```python
#right join
pd.merge(df_customers, df_orders, on="customer_id", how="right")
```

```python
#outer join
pd.merge(df_customers, df_orders, on="customer_id", how="outer")
```

#### Data Concatenations with Pandas
