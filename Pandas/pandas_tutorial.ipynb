### Series in Pandas

```python
s = pd.Series([1, 2, 3, 4, 5])
print(s)
print(type(s))

#index 
print(s[0])
print(s[2])
```

```python
# label
s = pd.Series([23, 24, 25, 26], index = ["Adam", "Eve", "Bob", "charlie"])
print(s)

print(s["Adam"])
print(s["Bob"])
print(s.index)
```

```python
# Series Properties
s = pd.Series([23, 24, "abc"], index = ["Adam", "Eve", 25])  # Homogeneous 
print(s)

# -> Vectorized Operations
s1 = pd.Series([1, 2, 3, 4, 5])       
s2 = pd.Series([10, 20, 30, 40, 50])
print(s1 + s2)     

s1[0] = 100
print(s1)
changed_s1 = s1.drop(0)
print(changed_s1)
```

```python
# DataFrame

info = {
    "Name" : ["Adam", "Bob", "Eve"],
    "Age" : [23, 24, 25],
    "GPA" : [9.5, 8.6, 7.3]
}

df = pd.DataFrame(info)

print(df, type(df))

print(df.index)
print(df.columns)
```

```python
# list by DataFrame
df = pd.DataFrame([["Adam", 23], ["Bob", 24], ["Eve", 23]], columns = ["Name", "Age"])
print(df)
```

```python
# numpy by DataFrame
import numpy as np

np_arr = np.array([[1, 2, 3], [4, 5, 6]])

df = pd.DataFrame(np_arr, columns = ["A", "B", "C"])

print(df)
```

```python
# csv data
df = pd.read_csv("offers-1000.csv")
print(df, type(df))
```
