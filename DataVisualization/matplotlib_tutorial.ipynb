# Matplotlib

```python
import matplotlib.pyplot as plt
```

```python
X = [1, 2, 3, 4]
Y = [5, 6, 7, 8]

plt.plot(X, Y) #line plot
plt.show()
```

```python
# Real data

# Oscar Winner Movies
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The kings's Speech",
    "The Artist",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]  #in $M

# Plot
plt.plot(years, oscar_revenue)

plt.title("Oscar Movies Revenue in each Year")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
```

```python
# Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionaire", "Avatar", "Inception", "Hugo", "Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275]  # in $M

plt.plot(years, oscar_revenue)
plt.plot(years, non_oscar_revenue)

plt.title("Oscar Movies vs Non-Oscar Movies Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")
```

```python
## label & legend
plt.plot(years, oscar_revenue, label="Oscar Movies")
plt.plot(years, non_oscar_revenue, label="Non-Oscar Movies")

plt.title("Oscar Movies vs Non-Oscar Movies Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")

plt.legend()
# plt.legend(loc="center")
# plt.legend(loc="upper left")
# plt.legend(loc="lower left")
```

## Format Strings

```python
# fmt= "[marker][line][color]"
plt.plot(years, oscar_revenue, "o--r", label="Oscar Movies")
plt.plot(years, non_oscar_revenue, "<-y",label="Non-Oscar Movies")

plt.title("Oscar Movies vs Non-Oscar Movies Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")

plt.legend()
```

```python
# fomat strings:- color, marker, linestyle, linewidth
plt.plot(years, oscar_revenue, color="brown", marker="<", linestyle="-.", linewidth=5, label="Oscar Movies")
plt.plot(years, non_oscar_revenue, label="Non-Oscar Movies")

plt.title("Oscar Movies vs Non-Oscar Movies Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")

plt.legend()
```
